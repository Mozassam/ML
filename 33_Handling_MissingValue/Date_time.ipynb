# Working with date

```python
# converting the object date in datetime datatype
date['date']=pd.to_datetime(date['date'])
```

```python
date.info()
```

<h2>Extracting year</h2>

```python
date['date_year']=date['date'].dt.year
```

```python
date.head()
```

<h2>Extracting Month</h2>

```python
date['date_month']=date['date'].dt.month
date.head()
```

```python
date['date_month_name']=date['date'].dt.month_name()
```

```python
date.head()
```

<h2>Extracting day</h2>

```python
date['date_day']=date['date'].dt.day
date.head()
```

```python
date['date_day_ofweek']=date['date'].dt.dayofweek
date.head()
```

```python
date['date_day_name']=date['date'].dt.day_name()
date.head()
```

```python
# is weekend or not?
date['is_weekend']=np.where(date['date_day_name'].isin(['Saturday','Sunday']),1,0)  # we can write true,false in place of 1,0
date.head()
```

```python
# weak of the year
date['week_no'] = date['date'].dt.isocalendar().week
date.head()
```

Time between two dates


```python
import datetime as dt
```

```python
tdate = dt.datetime.today()
tdate
```

```python
tdate-date['date']
```

```python
# if you want just days then
(tdate-date['date']).dt.days
```

```python
# Months passed

months_passed = (
    (tdate.year - date['date'].dt.year) * 12
    + (tdate.month - date['date'].dt.month)
)

print(months_passed)
```
