```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.impute import SimpleImputer, KNNImputer
```

```python
df=pd.read_csv(r"C:\Users\Mozassam Imam\Desktop\AI\CampusX\ML\19_Understanding_data\train_titanic.csv",usecols=['Age','Pclass','Fare','Survived'])
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
X=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
X_train,X_test,y_train,y_test= train_test_split(X,y,test_size=0.2, random_state=2)
```

```python
X_train
```

```python
knn = KNNImputer(n_neighbors=3,weights='distance')
```

```python
X_train_trf=knn.fit_transform(X_train)
X_test_trf=knn.transform(X_test)
```

```python
lr = LogisticRegression()
```

```python
lr.fit(X_train_trf,y_train)
y_pred=lr.predict(X_test_trf)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_test,y_pred)
```

```python
si = SimpleImputer()
```

```python
X_train_trf1=si.fit_transform(X_train)
X_test_trf1=si.transform(X_test)

lr.fit(X_train_trf1,y_train)

y_pred1 = lr.predict(X_test_trf1)
accuracy_score(y_test,y_pred1)
```

```python
def knn_at_every_value(X_train,X_test,y_train,y_test):
    for i in range (1,20):
        knn = KNNImputer(n_neighbors=i,weights='distance')
        X_train_trf=knn.fit_transform(X_train)
        X_test_trf=knn.transform(X_test)
        lr.fit(X_train_trf,y_train)
        y_pred=lr.predict(X_test_trf)
        print(f"accuracy when neighbors is {i} ",accuracy_score(y_test,y_pred))
```

```python
knn_at_every_value(X_train,X_test,y_train,y_test)
```

```python
def knn_at_every_value1(X_train,X_test,y_train,y_test):
    neighbors=[]
    accuracies=[]
    for i in range (1,21):
        knn = KNNImputer(n_neighbors=i,weights='distance')
        
        X_train_trf=knn.fit_transform(X_train)
        X_test_trf=knn.transform(X_test)
        
        lr.fit(X_train_trf,y_train)
        y_pred=lr.predict(X_test_trf)
        
        accuracy = accuracy_score(y_test, y_pred)
        
        neighbors.append(i)
        accuracies.append(accuracy)
    result=pd.DataFrame({
    "n_neighbors":neighbors,
    "accuracy":accuracies
    })
    return result
```

```python
knn_at_every_value1(X_train,X_test,y_train,y_test)
```
