```python
import http.client

conn = http.client.HTTPSConnection("f1-live-pulse.p.rapidapi.com")

headers = {
    'x-rapidapi-host': "f1-live-pulse.p.rapidapi.com",
    'Content-Type': "application/json"
}

conn.request("GET", "/fiaDocuments", headers=headers)

res = conn.getresponse()
data = res.read()

print(data.decode("utf-8"))
```

```python
import http.client

conn = http.client.HTTPSConnection("instagram-looter2.p.rapidapi.com")

headers = {
    'x-rapidapi-host': "instagram-looter2.p.rapidapi.com",
    'Content-Type': "application/json"
}

conn.request("GET", "/search?query=us", headers=headers)

res = conn.getresponse()
data = res.read()

print(data.decode("utf-8"))
```
